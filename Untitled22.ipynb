```python
# PYTHON PROGRAM — Student Report Generator

# student = "CHIPRA-PYTHON-2026"
# marks = [78, 91, 84, 67, 88]
#
# Using Python, create a small program that prints:
#
# Student: CHIPRA
# Subject: PYTHON
# Year: 2026
# Marks: [78, 91, 84, 67, 88]
# Highest marks: 91
# Lowest marks: 67

# Requirements:

student = "CHIPRA-PYTHON-2026"
marks = [78, 91, 84, 67, 88]

# 1. Use slicing to extract "CHIPRA".
print(student[0:6])
# 2. Use slicing to extract "PYTHON".
print(student[7:13])
# 3. Use slicing to extract "2026".
print(student[14:18])
# 4. Print the marks.
print(marks)
# 5. Find and print the highest mark.
print("Highest Marks:",marks[1] )
# 6. Find and print the lowest mark.
print("Lowest marks:",marks[3])
# 7. Extract the last 6 characters using negative indexing.
print(student[-6::])
#8. Reverse "PYTHON" using slicing.
print(student[-11:-5])
#9. Reverse the complete student record.
print(student[::-1])
#10. Extract every second character, first starting from
#    index 0 and then starting from index 1.
print(student[0::2])
print(student[1::2])
#11. Create a backward slice using negative indexes that
#    extracts "NOHTYP".
print(student[-6:-12:-1])
#12. Create one empty slice because the direction is impossible,
#    then change ONLY the step to make it non-empty.
print(student[-5:-7])
print(student[-5:-7:-1])
#13. Using the marks list:
#    - extract the last 3 marks using negative indexing,
print(marks[-3::])
#    - reverse the list using slicing,
print(marks[::-1])
#    - create a complete copy using slicing.
copy= marks[:]
print("copy:",copy)


```
