```python
#Write a Python program that takes a sentence from the user and displays:
sentence = input("Enter a sentence:")
#1. The original sentence
print(sentence)
#2. The sentence in UPPERCASE
print(sentence.upper())
#3. The sentence in lowercase
print(sentence.lower())
#4. The sentence with the first character capitalized
print(sentence.capitalize())
#5. The sentence with Every Word's First Letter Capitalized
print(sentence.title())
#6. The sentence with uppercase letters changed to lowercase and lowercase letters changed to uppercase
print(sentence.swapcase())
```
