```python
#Program 7 — Text Correction Tool

#Write a program that completes these tasks.
message = "chipra loves java. chipra is learning java programming."
#Part A — Replacing strings
#1. Replace "java" with "python" throughout the message.
message = message.replace("java", "Python")
#2. Replace "chipra" with "Chipra" throughout the message.
message = message.replace("chipra", "Chipra")
#3. Replace only the first occurrence of "java" with "Python".
message = message.replace("java", "Python",1)
#4. Replace "programming" with "coding".
message = message.replace("programming", "coding")
#Part B — Finding and checking
#5. Find the first occurrence of "java".
print(message.find("java"))
#6. Count how many times "java" appears.
print(message.count("java"))
#7. Check whether the original message starts with "chipra".
print (message.startswith("chipra"))
#8. Check whether the original message ends with "programming.".
print(message.endswith("programming"))
#Part C — Changing case
#9. Display the original message in uppercase.
print(message.upper())
#10. Display the original message with its case swapped.
print(message.swapcase())
#Part D — Slicing revision
#11. Display the first six characters.
print(message[0:6])
#12. Display the last eleven characters.
print(message[-1:-12:-1])
#13. Reverse the entire message.
print(message[::-1])
#14. Display every second character, starting from the first.
print(message[::2])
```
