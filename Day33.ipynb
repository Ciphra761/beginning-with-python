```python

# Program 9 — Whitespace Cleaner + Slicing Practice

text = "   chipra LOVES python programming   "
text1= "   chipra LOVES python programming   "
text2=  "   chipra LOVES python programming   "

# Part A — Removing Whitespace
# 1. Remove spaces from both ends and print the result using strip().
print(text.strip())
# 2. Remove only the spaces from the beginning and print the result using lstrip().
print(text1.lstrip())
# 3. Remove only the spaces from the end and print the result using rstrip().
print(text2.rstrip())
# 4. Remove spaces from both ends and save the updated string back into text.
text = text.strip()
# 5. Print the length of the updated text using len().
print(len(text))
# Part B — Slicing Revision
# 6. Print the first 6 characters of text.
print(text[0:6])
# 7. Print the last 11 characters in normal order.
print(text[-11::])
# 8. Print the last 5 characters in reverse order.
print(text[-1:-6:-1])
# 9. Print every second character, starting from the first.
print(text[::2])
# 10. Reverse the entire text and print it.
print(text[::-1])

# Important:
# - Use print() for every requested output.
# - For Questions 1–3, print each method's result separately.
# - For Question 4, use reassignment.

```
