```python

# Program 10 — Text Inspection Challenge

message = "   chipra loves Python. Python makes coding fun!   "

# Part A — Removing Whitespace (Set 4)
# 1. Remove spaces from both ends and print the result using strip().
print(message.strip())
# 2. Remove only the left-side spaces and print the result using lstrip().
print(message.lstrip())
# 3. Remove only the right-side spaces and print the result using rstrip().
print(message.rstrip())
# 4. Remove spaces from both ends and reassign the cleaned string to message.
message = message.strip()
# 5. Print the length of the cleaned message using len().
print(len(message))

# Part B — String Methods (previously learned)
# 6. Count how many times "Python" appears in message.
print(message.count("Python"))
# 7. Find the first occurrence of "coding" using find().
print(message.find("coding"))
# 8. Check whether message starts with "chipra".
print(message.startswith("chipra"))
# 9. Check whether message ends with "fun!".
print(message.endswith("fun!"))

# Part C — Slicing
# 10. Print the first 6 characters.
print(message[0:6])
# 11. Print the last 4 characters in normal order.
print(message[-1:-5])
# 12. Print the last 5 characters in reverse order.
print(message[-1:-6])
# 13. Print every second character, starting from the first.
print(message[::2])
# 14. Reverse the entire cleaned message.
print(message[::-1])

# Part D — Replacement
# 15. Replace "Python" with "Java" throughout message and print the result.
print(message.replace("Python","Java"))
```
