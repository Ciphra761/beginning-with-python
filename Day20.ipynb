```python
#🐍 Program 3 — Slicing Challenge
message = "PYTHON-CAO-2026-NET-AI"

#Using slicing only:

#Extract CAO and print it backwards → OAC
print(message[-13:-16:-1])
#Extract 2026 and print it backwards → 6202
print(message[-8:-12:-1])
#Print the entire message backwards.
print(message[::-1])
#Print every second character starting from index 0.
print(message[::2])
#Print every second character starting from index 1.
print(message[1::2])
#Create a backward slice that begins inside 2026 and crosses backward into CAO.
print(message[14:6:-1])
#Rewrite #6 using negative indexes, producing exactly the same result.
print(message[-8:-16:-1])
#Create an empty slice caused by the direction of the step, then change only the step so that it produces a non-empty result.
print(message[9:10])

#That's it.
```
