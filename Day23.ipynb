```python
#🐍 Program 2 — Name Formatter
#Write a program that asks the user for their full name and displays:
name = input("Enter your name: ")
#1. Original name
print(name)
#2. Name in UPPERCASE
print(name.upper())
#3. Name in lowercase
print(name.lower())
#4. Name with only the first character capitalized
print(name.capitalize())
#5. Name with Every Word Capitalized
print(name.title())
#6. Name with uppercase letters changed to lowercase and vice versa
print(name.swapcase())
print(name[0:7])
print(name[7::])
print(name[::-1])
```
