```python

# Program 23: Practice the partition() Method

# 1. Take a name as input.
#    Store it in a variable named sentence.
sentence = input("Enter a name")
# 2. Partition the sentence using the separator "is".
#    Store the result in a variable named result1.
#    Display result1.
result1 = sentence.partition("is")
print(result1)
# 3. Take a filename as input.
#    Partition it using ".".
#    Store the result in result2.
#    Display result2.
filename = input("ENter the filename")
result2 = filename.partition(".")
print(result2)

# 4. Take a word as input.
#    Partition it using its first letter.
#    Store the result in result3.
#    Display result3.
word = input("Enter the word")
result3 = word.partition(word[0])
print(result3)

# 5. Take a string as input.
#    Partition it using "@".
#    Store the result in result4.
#    Display result4.
string = input("Enter a sentence")
result4 = string.partition("@")
print(result4)
```
