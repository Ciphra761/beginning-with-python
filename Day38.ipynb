```python

# Program 22: Exploring the partition() Method

# partition() divides a string into three parts:
# 1. Text before the separator
# 2. The separator itself
# 3. Text after the separator

# Case 1: Separator in the middle
text = "Python is fun"
result = text.partition("is")
print("Case 1:", result)

# Case 2: Separator at the beginning
text = "@chipra"
result = text.partition("@")
print("Case 2:", result)

# Case 3: Separator at the end
text = "Python!"
result = text.partition("!")
print("Case 3:", result)

# Case 4: Separator does not exist
text = "Python is fun"
result = text.partition("-")
print("Case 4:", result)

# Case 5: Separator occurs multiple times
# partition() uses only the first occurrence.
text = "banana"
result = text.partition("a")
print("Case 5:", result)
```
