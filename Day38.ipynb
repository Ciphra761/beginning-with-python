```python
# PROGRAM 20: STRING METHODS CHALLENGE

# 1. Ask the user to enter a sentence.
sentence = input("enter a sentence:")
# 2. Remove leading and trailing spaces.
#    Store the result in a new variable.
cleaned = sentence.strip()
# 3. Display the total number of characters.
print(len(cleaned))
# 4. Display the sentence in title case and swapcase.
print(cleaned.title())
print(cleaned.swapcase())
# 5. Ask the user to enter a word.
word = input("Enter a word")
#    Display its first position and occurrence count in the sentence.
print(cleaned.find(word))
# 6. Check whether the sentence starts with "Python".
print(cleaned.startswith("Python"))
# 7. Check whether the sentence contains only letters and numbers.
print(cleaned.isalnum())
# 8. Replace the entered word with "Programming".
updated = cleaned.replace(word,"Programming")
#    Store and display the updated sentence.
print(updated)
# 9. Display the first five characters of the updated sentence.
print(updated[0:5])
# 10. Display the last five characters in reverse order.
print(updated[-1:-6:-1])
# END OF PROGRAM
```
