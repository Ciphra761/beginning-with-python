```python
# PROGRAM 21: USERNAME VALIDATOR

# 1. Ask the user to enter a username.
username = input("Enter your username:")
# 2. Remove leading and trailing spaces.
#    Store the cleaned username in a new variable.
cleaned = username.strip()
# 3. Check whether the username contains only letters and numbers.
print(username.isalnum())
# 4. Display the username in lowercase.
print(username.lower())
# 5. Check whether the username starts with "Chipra".
print(username.startswith("Chipra"))
# 6. Count how many times the letter "a" occurs.
print(username.count("a"))
# 7. Ask the user for a character to replace.
char = input("Input any letter:")
#    Replace it with "X" and store the updated username.
updated = username.replace(char,"X")
# 8. Display the first three characters of the updated username.
print(updated[0:3])
# 9. Display the total number of characters in the updated username.
print(len(updated))
# END OF PROGRAM
```
