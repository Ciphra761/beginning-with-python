```python

# Program 13 — String Classification Challenge

# 1. Check whether "ChipraVale" contains only alphabets.
print("ChipraVale".isalpha())
# Output: True


# 2. Check whether "20261103" contains only digits.
print("20261103".isdigit())
# Output: True


# 3. Check whether "Python2026" contains only letters and numbers.
print("Python2026".isalnum())
# Output: True


# 4. Check whether "hello python" contains only lowercase letters.
print("hello python".islower())
# Output: True


# 5. Check whether "PYTHON 2026!" contains only uppercase letters.
print("PYTHON 2026!".isupper())
# Output: True


# 6. Check whether the string contains only whitespace.
print("     ".isspace())
# Output: True


# 7. Check whether the string is in title case.
print("Learning python is fun".istitle())
# Output: False


# 8. Check whether the username contains only letters and numbers.
username = "Chipra_25"
print(username.isalnum())
# Output: False
# Reason: Underscore is neither a letter nor a number.


# 9. Check whether the password contains only alphabetic characters.
password = "Python@2026"
print(password.isalpha())
# Output: False
# Reason: It contains a special character and numbers.


# 10. Check whether all cased letters are lowercase and uppercase.
sentence = "I LOVE PYTHON"

print(sentence.islower())
# Output: False

print(sentence.isupper())
# Output: True
```
