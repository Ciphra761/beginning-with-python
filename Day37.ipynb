```python
# Program 16 — Student Profile Analyzer

# 1. Ask the user to enter their full name.
#    Remove any extra spaces from the beginning and end.
#    Display the cleaned name.
name = input("Enter your name :")
cleanname = name.strip()
print(cleanname)
# 2. Check whether the cleaned name contains only alphabets and spaces.
#    (Hint: You may need to check each word separately later.
#    For now, use a suitable method on the full name.)
print("does name contains only alphabets:")
print(cleanname.isalpha())
print(cleanname.isspace())

# 3. Display the name in uppercase and title case.
print(name.upper())
print(name.title())

# 4. Ask the user to enter their favourite programming language.
#    Check whether it starts with "Py".
pl = input("Enter your favourite programming language :")
print("Lets check if the language starts with py")
print(pl.startswith("Py"))

# 5. Count how many times the letter "a" appears in the language name.
print("In the name the letter occurs ___ times ")
print(pl.count("a"))

# 6. Display the first three characters of the language name.
print("first three characters of the names")
print(pl[0:3])

# 7. Ask the user to enter a short introduction.
#    Check whether it ends with a full stop.
sentences = input("Kindly give your short introduction:")
print(sentences.endswith("."))

# 8. Replace every occurrence of "Python" with "Programming"
#    in the introduction and display the updated introduction.
print(sentences.replace("Python","Programming"))
```
