```python
# PROGRAM 19: PERSONAL TEXT ANALYZER

# 1. Ask the user to enter their full name.
name = input("Enter your full name:")
# 2. Remove unnecessary spaces from both ends.
#    Store the cleaned name in a new variable.
cleaned = name.strip()
# 3. Check whether the cleaned name contains only alphabets.
#    Display the result using .isalpha().
print(cleaned.isalpha())

# 4. Ask the user to enter a word to search for.

# 5. Check whether the word exists in the cleaned name.
#    Use .find() and display its first position.
word = input("Enter a word: ")
print(cleaned.find(word))
# 6. Count how many times the word occurs.
#    Display the result using .count().
print(cleaned.count(word))

# 7. Convert the cleaned name and search word into lowercase.
#    Store both lowercase versions in separate variables.
cleanedL = cleaned.lower()
wordL = word.lower()

# 8. Search for the lowercase word in the lowercase name.
#    Display its first position using .find().
print(cleanedL.find(wordL))

# 9. Replace the search word in the cleaned name with "Python".
#    Store the updated name in a new variable and display it.
cleaned = cleaned.replace(word,"Python")
print(cleaned)
# 10. Display the last five characters of the updated name
#     in reverse order using slicing.
print(cleaned[-1:-6:-1])

# END OF PROGRAM
```
