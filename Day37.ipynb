```python
# PROGRAM 18: CASE-SENSITIVE SEARCH LAB

# 1. Ask the user to enter a sentence.
sentence = input("  Enter a sentence:  ")
# 2. Remove unnecessary spaces from both ends using .strip().
#    Store the cleaned sentence in a new variable.
cleaned = sentence.strip()
# 3. Ask the user to enter a word to search for.
word = input("Enter the word: ")
# 4. Use .find() to display the first position of the word.
print(sentence.find(word))
# 5. Use .count() to display how many times the word occurs.
print(sentence.count(word))
# 6. Use .index() to find the first position of the word.
#    Observe what happens if the word is not found.
print(sentence.index(word))
# 7. Create two new variables:
#    - Store the lowercase version of the cleaned sentence.
#    - Store the lowercase version of the search word.
cleaned1 = cleaned.lower()
word1 = word.lower()
#    Use these two new variables with .find()
#    to find the first occurrence of the word.
position = cleaned1.find(word1)

#    Display the result.
print(position)

# 8. Display the last five characters of the cleaned sentence
#    in reverse order using slicing.
print(cleaned[-1:-6:-1])

# 9. Check whether the cleaned sentence ends with a full stop
#    using .endswith().
print(cleaned.endswith("."))

# END OF PROGRAM
```
