```python
# Program 17 — Mini Text Data Analyzer

# 1. Ask the user to enter a sentence.
sentence = input("Enter the sentence:")
#    Remove extra spaces from both ends.
#    Store and display the cleaned sentence.
cleaned = sentence.strip()
print(cleaned)
# 2. Display the total number of characters in the cleaned sentence.
print(len(cleaned))

# 3. Display the sentence in title case and swapcase.
print(sentence.title())
print(sentence.swapcase())

# 4. Ask the user to enter a word to search for.
word = input("Enter a word")
#    Display its first position in the sentence.
print(sentence.find(word))
#    Also display how many times it occurs.
print(sentence.count(word))

# 5. Check whether the sentence starts with "Python".
print(sentence.startswith("Python"))
# 6. Check whether the sentence ends with a full stop.
print(sentence.endswith("."))
# 7. Ask the user to enter a word to replace.
replaceword = input("Enter the word to replace:")

#    Replace that word with "Programming".
updated =(sentence.replace(replaceword,"programming"))
#    Display the updated sentence.
print(updated)
# 8. Display the last five characters of the updated sentence
#    in reverse order.
print(sentence[-1:-6:-1])
```
