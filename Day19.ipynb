```python
code = "PY3-CAO_2026_NET-AIR1"
# Build a program that uses slicing only to extract and print:

#PY3
print(code[0:3])

#CAO
print(code[4:7])

#2026
print(code[8:12])

#NET
print(code[13:16])

#AIR1
print(code[17:21])

#The complete code backwards
print(code[::-1])

#CAO backwards
print(code[6:3:-1])

#Every second character of the complete code, starting from index 0
print(code[::2])

#Every second character of the complete code, starting from index 1
print(code[1::2])

#The last 8 characters, using a negative index
print(code[-8:])

#Everything except the first 4 characters
print(code[4:])

#Everything except the last 5 characters
print(code[0:-5])

#A slice whose stop is deliberately beyond the string length
print(code[0:20])

#An empty slice where start == stop
print(code[0:0])

#A backward slice that starts somewhere in the middle and stops before the beginning
print(code[10:0:-1])

#The same backward slice as #15, but using negative indexes instead of positive indexes.
print(code[-11:-21:-1])
```
