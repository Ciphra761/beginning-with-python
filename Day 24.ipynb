```python
#🐍 Program 4 — Secret Agent Profile
#Write a Python program that asks the user to enter a secret agent's name.
name= "Chipra Vale"
#Part A — String methods
#Your program must display:
#1. The original name.
print(name)
#2. The name in uppercase.
print(name.upper())
#3. The name in lowercase.
print(name.lower())
#4. The name with only the first character capitalized.
print(name.capitalize())
#5. The name with every word capitalized.
print(name.title())
#6. The name with its case swapped.
print(name.swapcase())
#Part B — Slicing revision
#7. Display the first 6 characters of the name.
print(name[0:7])
#8. Display the last 4 characters of the name.
print(name[-4::])
#9. Display the name backwards.
print(name[::-1])
#10. Display every second character, starting from the first character.
print(name[::2])
```
