```python
# TYPE CONVERSION — SHOPPING & TRAVEL BUDGET
# 1. Ask for:
#    - Name
#    - Available money (decimal)
#    - Price of 3 items (decimal)
#    - Quantity of each item (integer)
# 2. Calculate:
#    - Cost of each item
#    - Total shopping cost
#    - Money left after shopping
# 3. Ask for:
#    - Travel days (integer)
#    - Daily travel budget (decimal)
# 4. Calculate:
#    - Total travel budget
#    - Total expenditure
#    - Final money remaining
# 8. Print a neat final Trip Budget Report.

name=input("Enter your name: ")
money=float(input("Enter your available money: "))
item1=float(input("Enter the price of item 1: "))
item2=float(input("Enter the price of item 2: "))
item3=float(input("Enter the price of item 3: "))
quantity1=int(input("Enter the quantity of item 1: "))
quantity2=int(input("Enter the quantity of item 2: "))
quantity3=int(input("Enter the quantity of item 3: "))
cost_item1=item1*quantity1
cost_item2=item2*quantity2
cost_item3=item3*quantity3
totalcost=cost_item1+cost_item2+cost_item3
moneyleft=money-totalcost
travel_days=int(input("Enter the number of travel days: "))
daily_travel_budget=float(input("Enter your travel budget for each day: "))
total_travel_budget=daily_travel_budget*travel_days
final_money_remaining=moneyleft-total_travel_budget
print("Trip Budget Report")
print("travel_days: ",travel_days)
print("daily travel budget: ",daily_travel_budget)
print("total travel budget: ",total_travel_budget)
print("final money remaining: ",final_money_remaining)
```
