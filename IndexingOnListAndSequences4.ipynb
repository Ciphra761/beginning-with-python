```python
numbers = [10,20,30,40,50]
print("First number:",numbers[0])
print("Second number:",numbers[1])
print("Third numbers:",numbers[2])
print("Fourth numbers:",numbers[3])
print("Fifth numbers:",numbers[4])
print("Last number:",numbers[-1])
print("Second last number:",numbers[-2])
print("Third last number:",numbers[-3])
print("Fourth last number:",numbers[-4])
print("Fifth last number:",numbers[-5])
index=2
print("Number at index",index,":", numbers[index])

```
