```python
word = "PYTHON"
# 1. Reverse the entire string
print(word[::-1])

# 2. Reverse using a negative step
print(word[5:1:-1])

# 3. Slice using negative indexes
print(word[-1:-4:-1])

# 4. Reverse from a particular postion
print(word[4::-1])

# 5. Reverse a portion of the string
print(word[3:0:-1])


```
