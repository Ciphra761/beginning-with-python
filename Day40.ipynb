```python
# 1. Take a word as input.
#    Use center() with width 15 and "*" as the fill character.
#    Store the result in result1 and display it.
word = input("Enter the word")
result1 = word.center(15,"*")
print(result1)
# 2. Take a name as input.
#    Use ljust() with width 12 and "-" as the fill character.
#    Store the result in result2 and display it.
name = input("Enter your name:")
result2 = name.ljust(12,"-")
print(result2)
# 3. Take a city as input.
#    Use rjust() with width 18 and "#" as the fill character.
#    Store the result in result3 and display it.
city = input("Enter your city name")
result3 = city.rjust(18,"#")
print(result3)
```
