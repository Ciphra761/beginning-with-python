```python

# Program 15 — User Input Validator

# 1. Ask the user to enter their name.
#    Check whether the name contains only alphabets.
name = input("Enter your name:")
print(name.isalpha())
# 2. Ask the user to enter their PIN.
#    Check whether the PIN contains only digits.
pin = input("Enter your PIN:")
print(pin.isdigit())
# 3. Ask the user to enter a username.
#    Check whether it contains only letters and numbers.
username = input("Enter your username:")
print(username.isalnum())
# 4. Ask the user to enter a sentence.
#    Check whether it is written in title case.
sentence = input("Enter your sentence:")
print(sentence.istitle())
# 5. Ask the user to enter a password.
#    Check whether all its cased letters are uppercase
#    and whether all its cased letters are lowercase.
password = input("Enter your password:")
print(password.isupper())
print(password.islower())
```
