```python

# Program 11 — Personal Text Editor

message = "   chipra loves Python and Python loves coding!   "

# Part A — Removing Whitespace (Set 4)
# 1. Print the message after removing spaces from both ends.
print(message.strip())
# 2. Print the message after removing only left-side spaces.
print(message.lstrip())
# 3. Print the message after removing only right-side spaces.
print(message.rstrip())
# 4. Remove spaces from both ends and reassign the result to message.
message = message.strip()
# 5. Print the length of the updated message using len().
print(len(message))

# Part B — Finding and Checking
# 6. Count how many times "Python" appears.
print(message.count("Python"))
# 7. Find the index of the first occurrence of "loves".
print(message.find("loves"))
# 8. Check whether message starts with "chipra".
print(message.startswith("chipra"))
# 9. Check whether message ends with "coding!".
print(message.endswith("coding!"))

# Part C — Replacement and Case
# 10. Replace "Python" with "Java" and print the result.
print(message.replace("Python","Java"))
# 11. Print the message in title case.
print(message.title())
# 12. Print the message with its case swapped.
print(message.swapcase())

# Part D — Slicing Practice
# 13. Print the first 6 characters.
print(message[0:6])
# 14. Print the last 8 characters in normal order.
print(message[-8::])
# 15. Print the last 6 characters in reverse order.
print(message[-1:-7:-1])
# 16. Print every second character, starting from the first.
print(message[::2])
# 17. Reverse the entire message.
print(message[::-1])
```
