```python
# 1. Take a word as input.
#    Use center() to align it within a width of 20.
#    Store the result in result1 and display it.
word = input("Enter any word")
result1 = word.center(20)
print(result1)
# 2. Take a name as input.
#    Use ljust() to align it to the left within a width of 15.
#    Use "-" as the fill character.
#    Store the result in result2 and display it.
name = input("Enter a name")
result2 = name.ljust(15,"-")
print(result2)
# 3. Take a city name as input.
#    Use rjust() to align it to the right within a width of 15.
#    Use "*" as the fill character.
#    Store the result in result3 and display it.
city = input("Enter a name as input")
result3 = city.rjust(15,"*")
print(result3)
```
