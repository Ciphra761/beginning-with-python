```python

# Program 24: Practice the rpartition() Method

# 1. Take a sentence as input.
#    Partition it using the separator "a".
#    Store the result in result1 and display it.
sentence = input("Enter a sentence :")
result1 = sentence.partition("a")
print(result1)
# 2. Take a filename as input.
#    Partition it using ".".
#    Store the result in result2 and display it.
filename = input("Enter the file name ")
result2 = filename.partition(".")
print(result2)

# 3. Take a word as input.
#    Partition it using its first letter.
#    Store the result in result3 and display it.
word = input("Enter a word")
result3 = word.partition(word[0])

# 4. Take a string containing the word "Python" multiple times.
#    RPartition it using "Python".
#    Store the result in result4 and display it.
string = input("Kindly enter a sample sentence which uses the word 'Python' multiple times")
result4 = string.partition("Python")
print(result4 )
# 5. Take a string containing multiple dots (.) as separators.
#    RPartition it using ".".
#    Store the result in result5 and display it.
string2 = input("Kindly enter a sample sentence which uses fullstops '.' multiple times")
result5 = string2.partition(".")
print(result5 )
```
