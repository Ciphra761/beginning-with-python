```python
# 1. Student ID
# Take a student ID as input.
# Pad it with zeros to make its total width 6.
# Store the result in student_id and display it.
studentid = input("Enter you student ID:")
student_id = studentid.zfill(6)
print(student_id)
# 2. Invoice number
# Take an invoice number as input.
# Pad it with zeros to make its total width 8.
# Store the result in invoice_id and display it.
invoicenumber = input("Your invoice number..")
invoice_id = invoicenumber.zfill(8)
print(invoice_id)
# 3. Product code
# Take a product code as input.
# Pad it with zeros to make its total width 10.
# Store the result in product_code and display it.
productcode = input("Enter your number:")
product_code =  productcode.zfill(0)
print(product_code)
# 4. Negative number
# Take a negative number as a string.
# Use zfill() to make its total width 7.
# Store the result in formatted_number and display it.
negativenumber = input("Enter the negative number")
formatted_number = negativenumber.zfill(7)
print(formatted_number)
```
