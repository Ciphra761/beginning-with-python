```python
#Demonstrating Arithemetic Operators and Operators Precedence
a = 10
b= 3
print("Addition:", a + b)
print("Substraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponent:", a ** b)
print("\n Lets document operator precedence")
print("10 + 5 * 2 ** 3 - 20 / 4 + 15 // 4 % 3 =",
      10 + 5 * 2 ** 3 - 20 / 4 + 15 // 4 % 3 )
```
