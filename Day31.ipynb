```python

## Program 6 — Library Book Inspector

book = "python Programming: Chipra Vale learns python programming."

### Part A — Finding and checking

# 1. Find the first occurrence of `"python"`.
print(book.find("python"))
# 2. Find the first occurrence of `"Programming"`.
print(book.find("Programming"))
# 3. Count how many times `"python"` appears.
print(book.count("python"))
# 4. Count how many times `"a"` appears.
print(book.count("a"))
# 5. Check whether the string starts with `"python"`.
print(book.startswith("python"))
# 6. Check whether the string ends with `"programming."`.
print(book.endswith("programming."))
# 7. Check whether `"Java"` exists in the string using `.find()`.
print(book.find("Java"))

### Part B — Changing case

#8. Display the entire string in uppercase.
print(book.upper())
#9. Display the entire string in title case.
print(book.title())
#10. Swap the uppercase and lowercase letters.
print(book.swapcase())

### Part C — Slicing revision

#11. Display the first 6 characters.
print(book[0:6])
#12. Display the last 12 characters.
print(book[-12::])
#13. Display the string backwards.
print(book[::-1])
#14. Display every second character, starting with the first.
print(book[0::2])
```
