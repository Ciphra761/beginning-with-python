```python
product = "LIPBALM-2026-MATTE-50G"

#Using slicing only, extract:
#1. LIPBALM
print(product[0:7])
#2. 2026
print(product[8:12])
#3. MATTE
print(product[13:18])
#4. 50G
print(product[19:22])
#5. Reverse the product name
print(product[6::-1])
#6. Reverse the entire product code
print(product[::-1])
#7. Extract the last two characters using negative indexing.
print(product[-2:])
#8. Extract the product information backwards from MATTE into 2026.
print(product[-5:-15:-1])

```
